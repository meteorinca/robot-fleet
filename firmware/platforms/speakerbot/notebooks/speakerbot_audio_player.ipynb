# SpeakerBot Audio Player Notebook

This notebook allows you to play **any audio file** (MP3, WAV, M4A, FLAC, OGG, AAC) located in this `notebooks/` folder directly through SpeakerBot's speaker over Wi-Fi.

---

### 1. Requirements & Setup
Requires `requests` and `pydub` (or `ffmpeg`). Run the cell below to install if needed:

```python
# !pip install requests pydub
```

### 2. Connect to SpeakerBot

```python
import os
from speakerbot_player import SpeakerBot

botnum = 1

# Set your SpeakerBot hostname or IP address
SPEAKERBOT_HOST = f"http://speakerbot{botnum}.local"

bot = SpeakerBot(host=SPEAKERBOT_HOST)

if bot.ping():
    print(f"✅ Connected to SpeakerBot at {SPEAKERBOT_HOST}")
    print("Status:", bot.get_status())
else:
    print(f"⚠️ Unable to reach SpeakerBot at {SPEAKERBOT_HOST}. Check WiFi / device hostname.")
```

### 3. Option A: Synchronous Play (Pressing Interrupt / Stop Kernel Silences Hardware)
Pass any audio file. Pressing **Interrupt Kernel** (⏹ in VSCode/Jupyter) will automatically silence SpeakerBot immediately.

```python
files = [f for f in os.listdir('.') if f.lower().endswith(('.mp3', '.wav', '.m4a', '.ogg', '.flac'))]
print("🎵 Available audio files:", files)

if files:
    target_file = files[0]
    print(f"▶️ Playing: {target_file}")
    # volume=0.5 applies clean headroom to prevent hardware amplifier clipping
    bot.play_file(target_file, interrupt=True, volume=0.5)
else:
    print("💡 Drop any .mp3 or .wav file into this folder to test!")
```

### 4. Option B: Non-Blocking Background Stream (`play_file_async`)
Plays in a background thread so the cell finishes instantly. You can run `bot.stop()` in the next cell at any time!

```python
if 'files' in locals() and files:
    bot.play_file_async(files[0], interrupt=True, volume=0.5)
else:
    files = [f for f in os.listdir('.') if f.lower().endswith(('.mp3', '.wav', '.m4a', '.ogg', '.flac'))]
    if files:
        bot.play_file_async(files[0], interrupt=True, volume=0.5)
```

### 5. Stop / Silence SpeakerBot
Instantly flush playback buffers and silence SpeakerBot hardware:

```python
bot.stop()
```

### 6. Play Hardware Sound Clips
Play pre-loaded clips (`bark`, `paulbot`, `huh`, `yes`, `jump`, `ding`, `random`) with optional repeat:

```python
# Play dog bark 3 times with emergency preemption
bot.play_clip(name="bark", repeat=3, interrupt=True)
```
